```python
import random

expanded_rows = []

for category, templates in additional_templates.items():

    # Create 500 examples for each category
    for i in range(500):

        base_text = random.choice(templates)

        # Small realistic variations
        variations = [
            base_text,
            f"Paid for {base_text.lower()}",
            f"Payment for {base_text.lower()}",
            f"Monthly {base_text.lower()}",
            f"Expense for {base_text.lower()}",
        ]

        transaction = random.choice(variations)

        expanded_rows.append({
            "Transaction_Text": transaction,
            "Label": category
        })

expanded_df = pd.DataFrame(expanded_rows)

print("Expanded dataset shape:", expanded_df.shape)
print("\nCategory distribution:")
print(expanded_df["Label"].value_counts())
```

```python
# Combine original dataset with expanded categories

final_nlp_df = pd.concat(
    [
        train_df[["Transaction_Text", "Label"]],
        expanded_df[["Transaction_Text", "Label"]]
    ],
    ignore_index=True
)

print("Before cleaning:")
print("Shape:", final_nlp_df.shape)

print("\nCategory distribution:")
print(final_nlp_df["Label"].value_counts())
```

```python
def clean_transaction(text):
    text = str(text).lower()

    # Remove INR amounts
    text = re.sub(r'inr\s*\d+', ' ', text)

    # Remove reference numbers
    text = re.sub(r'ref\s*\d+', ' ', text)

    # Remove numbers
    text = re.sub(r'\d+', ' ', text)

    # Keep only letters and spaces
    text = re.sub(r'[^a-z\s]', ' ', text)

    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text)

    return text.strip()


final_nlp_df["Clean_Text"] = final_nlp_df["Transaction_Text"].apply(
    clean_transaction
)

final_nlp_df.head()
```

```python
before = len(final_nlp_df)

final_nlp_df = final_nlp_df.drop_duplicates(
    subset=["Clean_Text", "Label"]
).reset_index(drop=True)

after = len(final_nlp_df)

print("Rows before removing duplicates:", before)
print("Rows after removing duplicates:", after)
print("Duplicates removed:", before - after)
```

```python
X = final_nlp_df["Clean_Text"]
y = final_nlp_df["Label"]
```

```python
X_train, X_val, y_train, y_val = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Validation samples:", len(X_val))

print("\nTraining distribution:")
print(y_train.value_counts())

print("\nValidation distribution:")
print(y_val.value_counts())
```

```python
vectorizer = TfidfVectorizer(
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True
)

X_train_tfidf = vectorizer.fit_transform(X_train)
X_val_tfidf = vectorizer.transform(X_val)

print("Training TF-IDF shape:", X_train_tfidf.shape)
print("Validation TF-IDF shape:", X_val_tfidf.shape)
```

```python
from sklearn.svm import LinearSVC

svm_model = LinearSVC(
    class_weight="balanced",
    random_state=42
)

svm_model.fit(X_train_tfidf, y_train)

y_val_pred = svm_model.predict(X_val_tfidf)

print("Validation Accuracy:",
      accuracy_score(y_val, y_val_pred))

print("\nClassification Report:")
print(classification_report(y_val, y_val_pred))
```

```python
# Find validation examples that were classified incorrectly

errors_df = pd.DataFrame({
    "Transaction": X_val,
    "Actual": y_val,
    "Predicted": y_val_pred
})

errors_df = errors_df[
    errors_df["Actual"] != errors_df["Predicted"]
]

print("Number of errors:", len(errors_df))

print("\nMisclassified transactions:")
print(errors_df.to_string(index=False))
```

```python
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

labels = sorted(y.unique())

cm = confusion_matrix(
    y_val,
    y_val_pred,
    labels=labels
)

plt.figure(figsize=(12, 9))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=labels,
    yticklabels=labels
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Expense Category Confusion Matrix")
plt.show()
```
